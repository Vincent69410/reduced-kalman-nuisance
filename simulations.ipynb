```python
# Table I: true / announced standard deviations (max over cruise, U-turn, end)
import json
res = json.load(open('results2d.json'))
names = {'full': 'Full KF (n=8)', 'naive': 'Gamma=0, Q=Qx', 'infl': 'Gamma=0, inflated', 'good': 'Gamma(t), certified'}
print(f"{'filter':22s} {'vel cruise':>12s} {'vel turn':>12s} {'vel end':>12s} {'pos cruise':>12s} {'pos turn':>12s} {'pos end':>12s}")
for k in ['full', 'naive', 'infl', 'good']:
    r = res['rows'][k]
    cells_ = []
    for ph in ['cruise', 'turn', 'end']:
        tv, av, tp, ap = r[ph]
        cells_.append(f'{tv:.2f}/{av:.2f}')
    cells_v = cells_
    cells_p = []
    for ph in ['cruise', 'turn', 'end']:
        tv, av, tp, ap = r[ph]
        cells_p.append(f'{tp:.2f}/{ap:.2f}')
    print(f"{names[k]:22s} " + ' '.join(f'{c:>12s}' for c in cells_v + cells_p))
print('\nParameters:', res['params'])
print('\nScalar example:', json.load(open('results1d.json')))
```
